```python
# First Come First Serve (FCFS) Scheduling

n = int(input("Enter number of processes: "))

processes = []
burst_time = []

for i in range(n):
    processes.append(f"p{i+1}")
    bt = int(input(f"Enter burst time for {processes[i]}: "))
    burst_time.append(bt)

waiting_time = [0] * n

# Calculate waiting time
for i in range(1, n):
    waiting_time[i] = waiting_time[i-1] + burst_time[i-1]

# Calculate average waiting time
total_waiting_time = sum(waiting_time)
avg_waiting_time = total_waiting_time / n

# Generate Gantt Chart
print("\nGantt Chart:")
current_time = 0
print(current_time, end=" ")

for i in range(n):
    current_time += burst_time[i]
    print(processes[i], current_time, end=" ")

# Show waiting time calculation
print("\n\nWaiting Times:")
for i in range(n):
    print(f"{processes[i]} = {waiting_time[i]}")

# Show average waiting time calculation
print("\nAverage Waiting Time Calculation:")
print("= (", end=" ")
print(" + ".join(str(wt) for wt in waiting_time), end=" ")
print(f") / {n}")

print(f"= {total_waiting_time} / {n}")
print("= {:.2f}".format(avg_waiting_time))
```

```python
# Shortest Job First (SJF) - Non Preemptive

n = int(input("Enter number of processes: "))

processes = []
burst_time = []

# Auto assign process names and take burst time
for i in range(n):
    processes.append(f"p{i+1}")
    bt = int(input(f"Enter burst time for {processes[i]}: "))
    burst_time.append(bt)

# Combine process and burst time
combined = list(zip(processes, burst_time))

# Sort by burst time (Shortest Job First)
combined.sort(key=lambda x: x[1])

# Separate sorted values
sorted_processes = [x[0] for x in combined]
sorted_burst = [x[1] for x in combined]

# Calculate waiting time
waiting_time = [0] * n
for i in range(1, n):
    waiting_time[i] = waiting_time[i-1] + sorted_burst[i-1]

# Calculate average waiting time
total_waiting_time = sum(waiting_time)
avg_waiting_time = total_waiting_time / n

# Print Gantt Chart
print("\nGantt Chart:")
current_time = 0
print(current_time, end=" ")

for i in range(n):
    current_time += sorted_burst[i]
    print(sorted_processes[i], current_time, end=" ")

# Print Waiting Times
print("\n\nWaiting Times:")
for i in range(n):
    print(f"{sorted_processes[i]} = {waiting_time[i]}")

# Show Average Waiting Time Calculation
print("\nAverage Waiting Time Calculation:")
print("= (", end=" ")
print(" + ".join(str(wt) for wt in waiting_time), end=" ")
print(f") / {n}")
print(f"= {total_waiting_time} / {n}")
print("= {:.2f}".format(avg_waiting_time))

```

```python
# Priority Scheduling - Non Preemptive
n = int(input("Enter number of processes: "))
processes = []
burst_time = []
priority = []
for i in range(n):
  processes.append(f"p{i+1}")
  bt = int(input(f"Enter burst time for {processes[i]}: "))
  pr = int(input(f"Enter priority for {processes[i]} (lower number = higher priority): "))
  burst_time.append(bt)
  priority.append(pr)
# Combine all
datacombined = list(zip(processes, burst_time, priority))
# Sort by priority (lowest number first)
datacombined.sort(key=lambda x: x[2])
# Separate again after sorting
processes = [x[0] for x in datacombined]
burst_time = [x[1] for x in datacombined]
priority = [x[2] for x in datacombined]
waiting_time = [0] * n
# Calculate waiting time
for i in range(1, n):
  waiting_time[i] = waiting_time[i-1] + burst_time[i-1]
# Calculate average waiting time
total_waiting_time = sum(waiting_time)
avg_waiting_time = total_waiting_time / n
# Generate Gantt Chart
print("\nGantt Chart:")
current_time = 0
print(current_time, end=" ")
for i in range(n):
  current_time += burst_time[i]
  print(processes[i], current_time, end=" ")
# Show process details
print("\n\nProcess Details (After Priority Sorting):")
for i in range(n):
  print(f"{processes[i]} | Burst = {burst_time[i]} | Priority = {priority[i]}")
# Show waiting time calculation
print("\nWaiting Times:")
for i in range(n):
  print(f"{processes[i]} = {waiting_time[i]}")
# Show average waiting time calculation
print("\nAverage Waiting Time Calculation:")
print("= (", end=" ")
print(" + ".join(str(wt) for wt in waiting_time), end=" ")
print(f") / {n}")
print(f"= {total_waiting_time} / {n}")
print("= {:.2f}".format(avg_waiting_time))
```

```python
# Round Robin Scheduling

n = int(input("Enter number of processes: "))

processes = []
burst_time = []

for i in range(n):
    processes.append(f"p{i+1}")
    bt = int(input(f"Enter burst time for {processes[i]}: "))
    burst_time.append(bt)

time_quantum = int(input("Enter Time Quantum: "))

remaining_time = burst_time.copy()
waiting_time = [0] * n
completion_time = [0] * n

current_time = 0

# Round Robin Logic
while True:
    done = True

    for i in range(n):
        if remaining_time[i] > 0:
            done = False

            if remaining_time[i] > time_quantum:
                current_time += time_quantum
                remaining_time[i] -= time_quantum
            else:
                current_time += remaining_time[i]
                waiting_time[i] = current_time - burst_time[i]
                remaining_time[i] = 0
                completion_time[i] = current_time

    if done:
        break

# Calculate average waiting time
total_waiting_time = sum(waiting_time)
avg_waiting_time = total_waiting_time / n

# Generate Gantt Chart
print("\nGantt Chart:")
current_time = 0
print(current_time, end=" ")
for i in range(n):
  current_time += burst_time[i]
  print(processes[i], current_time, end=" ")

# Print Results
print("\nCompletion Times:")
for i in range(n):
    print(f"{processes[i]} = {completion_time[i]}")

print("\nWaiting Times:")
for i in range(n):
    print(f"{processes[i]} = {waiting_time[i]}")

print("\nAverage Waiting Time Calculation:")
print("= (", end=" ")
print(" + ".join(str(wt) for wt in waiting_time), end=" ")
print(f") / {n}")

print(f"= {total_waiting_time} / {n}")
print("= {:.2f}".format(avg_waiting_time))
```

```python
# FCFS Disk Scheduling algorithm

size = 8;

def FCFS(arr, head):

    seek_count = 0;
    distance, cur_track = 0, 0;

    for i in range(size):
        cur_track = arr[i];

        # calculate absolute distance
        distance = abs(cur_track - head);

        # increase the total count
        seek_count += distance;

        # accessed track is now new head
        head = cur_track;

    print("Total number of seek operations = ",
                                   seek_count);

    # Seek sequence would be the same
    # as request array sequence
    print("Seek Sequence is");

    for i in range(size):
        print(arr[i]);

# Driver code
if __name__ == '__main__':

    # request array
    arr = [98, 183, 37,
            122, 14,124, 65,67 ];
    head = 53;

    FCFS(arr, head);
```

```python
# SSTF disk scheduling

def calculateDifference(queue, head, diff):
    for i in range(len(diff)):
        diff[i][0] = abs(queue[i] - head)
def findMin(diff):

    index = -1
    minimum = 999999999

    for i in range(len(diff)):
        if (not diff[i][1] and
                minimum > diff[i][0]):
            minimum = diff[i][0]
            index = i
    return index

def shortestSeekTimeFirst(request, head):
        if (len(request) == 0):
            return

        l = len(request)
        diff = [0] * l

        # initialize array
        for i in range(l):
            diff[i] = [0, 0]

        seek_count = 0

        seek_sequence = [0] * (l + 1)

        for i in range(l):
            seek_sequence[i] = head
            calculateDifference(request, head, diff)
            index = findMin(diff)

            diff[index][1] = True

            # increase the total count
            seek_count += diff[index][0]

            # accessed track is now new head
            head = request[index]

        seek_sequence[len(seek_sequence) - 1] = head

        print("Total number of seek operations =",
                                       seek_count)

        print("Seek Sequence is")


        for i in range(l + 1):
            print(seek_sequence[i])

# Driver code
if __name__ =="__main__":

    # request array
    proc = [98, 183, 37, 122,
            14, 124, 65, 67]
    shortestSeekTimeFirst(proc, 53)
```

```python
# SCAN Disk Scheduling algorithm
size = 8
disk_size = 200

def SCAN(arr, head, direction):

    seek_count = 0
    distance, cur_track = 0, 0
    left = []
    right = []
    seek_sequence = []

    # Appending end values
    # which has to be visited
    # before reversing the direction
    if (direction == "left"):
        left.append(0)
    elif (direction == "right"):
        right.append(disk_size - 1)

    for i in range(size):
        if (arr[i] < head):
            left.append(arr[i])
        if (arr[i] > head):
            right.append(arr[i])

    # Sorting left and right vectors
    left.sort()
    right.sort()

    # Run the while loop two times.
    # one by one scanning right
    # and left of the head
    run = 2
    while (run != 0):
        if (direction == "left"):
            for i in range(len(left) - 1, -1, -1):
                cur_track = left[i]

                # Appending current track to
                # seek sequence
                seek_sequence.append(cur_track)

                # Calculate absolute distance
                distance = abs(cur_track - head)

                # Increase the total count
                seek_count += distance

                # Accessed track is now the new head
                head = cur_track

            direction = "right"

        elif (direction == "right"):
            for i in range(len(right)):
                cur_track = right[i]

                # Appending current track to seek
                # sequence
                seek_sequence.append(cur_track)

                # Calculate absolute distance
                distance = abs(cur_track - head)

                # Increase the total count
                seek_count += distance

                # Accessed track is now new head
                head = cur_track

            direction = "left"

        run -= 1

    print("Total number of seek operations =",
          seek_count)

    print("Seek Sequence is")

    for i in range(len(seek_sequence)):
        print(seek_sequence[i])

# Driver code

# request array
arr = [ 98,183,37,122,14,124,65,67]
head = 53
direction = "left"

SCAN(arr, head, direction)
```

```python
# C-SCAN Disk Scheduling algorithm
size = 8
disk_size = 200


def CSCAN(arr, head):

    seek_count = 0
    distance = 0
    cur_track = 0
    left = []
    right = []
    seek_sequence = []

    # Appending end values
    # which has to be visited
    # before reversing the direction
    left.append(0)
    right.append(disk_size - 1)

    # Tracks on the left of the
    # head will be serviced when
    # once the head comes back
    # to the beginning (left end).
    for i in range(size):
        if (arr[i] < head):
            left.append(arr[i])
        if (arr[i] > head):
            right.append(arr[i])

    # Sorting left and right vectors
    left.sort()
    right.sort()

    # First service the requests
    # on the right side of the
    # head.
    for i in range(len(right)):
        cur_track = right[i]

        # Appending current track
        # to seek sequence
        seek_sequence.append(cur_track)

        # Calculate absolute distance
        distance = abs(cur_track - head)

        # Increase the total count
        seek_count += distance

        # Accessed track is now new head
        head = cur_track

    # Once reached the right end
    # jump to the beginning.
    head = 0

    # adding seek count for head returning from 199 to 0
    seek_count += (disk_size - 1)

    # Now service the requests again
    # which are left.
    for i in range(len(left)):
        cur_track = left[i]

        # Appending current track
        # to seek sequence
        seek_sequence.append(cur_track)

        # Calculate absolute distance
        distance = abs(cur_track - head)

        # Increase the total count
        seek_count += distance

        # Accessed track is now the new head
        head = cur_track

    print("Total number of seek operations =",
          seek_count)
    print("Seek Sequence is")
    print(*seek_sequence, sep="\n")

# Driver code


# request array
arr = [98,183,37,122,14,124,65,67]
head = 50

print("Initial position of head:", head)

CSCAN(arr, head)
```

```python
# C-Look Disk Scheduling Algorithm
size = 8
disk_size = 200

# Function to perform C-LOOK on the request
# array starting from the given head
def CLOOK(arr, head):

    seek_count = 0
    distance = 0
    cur_track = 0

    left = []
    right = []

    seek_sequence = []

    # Tracks on the left of the
    # head will be serviced when
    # once the head comes back
    # to the beginning (left end)
    for i in range(size):
        if (arr[i] < head):
            left.append(arr[i])
        if (arr[i] > head):
            right.append(arr[i])

    # Sorting left and right vectors
    left.sort()
    right.sort()

    # First service the requests
    # on the right side of the
    # head
    for i in range(len(right)):
        cur_track = right[i]

        # Appending current track
        # seek sequence
        seek_sequence.append(cur_track)

        # Calculate absolute distance
        distance = abs(cur_track - head)

        # Increase the total count
        seek_count += distance

        # Accessed track is now new head
        head = cur_track

    # Once reached the right end
    # jump to the last track that
    # is needed to be serviced in
    # left direction
    seek_count += abs(head - left[0])
    head = left[0]

    # Now service the requests again
    # which are left
    for i in range(len(left)):
        cur_track = left[i]

        # Appending current track to
        # seek sequence
        seek_sequence.append(cur_track)

        # Calculate absolute distance
        distance = abs(cur_track - head)

        # Increase the total count
        seek_count += distance

        # Accessed track is now the new head
        head = cur_track

    print("Total number of seek operations =",
          seek_count)
    print("Seek Sequence is")

    for i in range(len(seek_sequence)):
        print(seek_sequence[i])

# Driver code

# Request array
arr = [98,183,37,122,14,124,65,67]
head = 50

print("Initial position of head:", head)

CLOOK(arr, head)
```

```python
# FIFO Page Replacement Algorithm

from collections import deque

frame_size = int(input("Enter frame size: "))
num_pages = int(input("Enter number of pages: "))

pages = list(map(int, input("Enter page reference string (space separated): ").split()))

frames = [-1] * frame_size
fifo_queue = deque()

page_faults = 0
page_hits = 0

print("\nPage Reference | Frames Status | Hit/Miss")
print("--------------------------------------------")

for current_page in pages:
    found = False


    if current_page in frames:
        found = True
        page_hits += 1

    print(f"      {current_page}        | ", end="")

    if not found:
        page_faults += 1


        if -1 in frames:
            empty_index = frames.index(-1)
            frames[empty_index] = current_page
            fifo_queue.append(current_page)
        else:

            page_to_replace = fifo_queue.popleft()
            replace_index = frames.index(page_to_replace)
            frames[replace_index] = current_page
            fifo_queue.append(current_page)

        # Print frames
        for f in frames:
            if f == -1:
                print("[ ]", end=" ")
            else:
                print(f"[{f}]", end=" ")

        print("| MISS")

    else:
        # Print frames
        for f in frames:
            if f == -1:
                print("[ ]", end=" ")
            else:
                print(f"[{f}]", end=" ")

        print("| HIT")


print("\n====== RESULTS ======")
print("Total Page Requests:", num_pages)
print("Total Page Hits:", page_hits)
print("Total Page Faults:", page_faults)
```

```python
# Optimal Page Replacement Algorithm


frame_size = int(input("Enter frame size: "))
num_pages = int(input("Enter number of pages: "))

pages = list(map(int, input("Enter page reference string (space separated): ").split()))

frames = [-1] * frame_size

page_faults = 0
page_hits = 0

print("\nPage Reference | Frames Status | Hit/Miss")
print("--------------------------------------------")

for i in range(num_pages):
    current_page = pages[i]
    found = False


    if current_page in frames:
        found = True
        page_hits += 1

    print(f"      {current_page}        | ", end="")

    if not found:
        page_faults += 1


        if -1 in frames:
            empty_index = frames.index(-1)
            frames[empty_index] = current_page
        else:

            future_uses = []

            for f in frames:
                if f in pages[i+1:]:
                    next_use = pages[i+1:].index(f)
                    future_uses.append(next_use)
                else:
                    future_uses.append(float('inf'))


            replace_index = future_uses.index(max(future_uses))
            frames[replace_index] = current_page

        # Print frames
        for f in frames:
            if f == -1:
                print("[ ]", end=" ")
            else:
                print(f"[{f}]", end=" ")

        print("| MISS")

    else:

        for f in frames:
            if f == -1:
                print("[ ]", end=" ")
            else:
                print(f"[{f}]", end=" ")

        print("| HIT")


print("\n====== RESULTS ======")
print("Total Page Requests:", num_pages)
print("Total Page Hits:", page_hits)
print("Total Page Faults:", page_faults)
```
